```python
import matplotlib.pyplot as plt
x=[10,15,20,25,30,35,40,45,50]
y=[1000,2000,3000,4000,5000,6000,7000,8000,9000]
plt.plot(x,y)
plt.show()
```

```python
x=[10,15,20,25,30,35,40,45,50]
y=[1000,2000,3000,4000,5000,6000,7000,8000,9000]
plt.figure(figsize=(8,5))
plt.plot(x,y,color='red',marker='o',linestyle='--',linewidth=2,markersize=8)
plt.title("comparision of age VS salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

```python
x=[1,3,4,5,6,7,8,9,12]
y1=[1000,2000,3000,4000,5000,6000,7000,8000,9000]
y2=[3000,5000,7000,9000,4000,7000,2000,10000,6000]
y3=[4000,6000,7000,1000,7000,3000,8000,6000,4000]
plt.plot(x,y1,y2,y3,color='red',marker='o',linestyle='--',linewidth=2,markersize=8)
plt.plot(x,y1,label='1st four month profit')
plt.plot(x,y2,label='2st four month profit')
plt.plot(x,y3,label='3st four month profit')
plt.title("comparision of sales profit")
plt.xlabel("Month")
plt.ylabel("Quater sales Profit")
plt.legend()
plt.show()
```

```python
x=[1,3,4,5,6,7,8,9,12]
y1=[1000,2000,3000,4000,5000,6000,7000,8000,9000]
y2=[3000,5000,7000,9000,4000,7000,2000,10000,6000]
y3=[4000,6000,7000,1000,7000,3000,8000,6000,4000]
plt.bar(x,y1,label='1st four month profit')
plt.bar(x,y2,label='2st four month profit')
plt.bar(x,y3,label='3st four month profit')
plt.title("comparision of sales profit")
plt.xlabel("Month")
plt.ylabel("Quater sales Profit")
plt.legend()
plt.show()
```
